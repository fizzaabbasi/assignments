Q4. You are managing a list of students enrolled in two different courses. 
    You need to perform various set operations to understand the student enrollment. 
    Task: Create two sets: one for students in "Course A" and one for students in "Course B". 
        Find the students who are in "Course A" but not in "Course B" (difference). 
        Find students who are only in one of the two courses (symmetric difference). 
            course_a = {"John", "Alice", "Bob", "David"} course_b = {"Alice", "Eve", "Charlie", "David"}

Using set methods

• Find students who are in Course A but not in Course B (difference)

• Find students who are only in one of the two courses (symmetric difference)


```python
course_a = {"John", "Alice", "Bob", "David"} 
course_b = {"Alice", "Eve", "Charlie", "David"}
diff = course_a.difference(course_b)
print(diff)
sym_diff = course_a.symmetric_difference(course_b)
print(sym_diff)
```

Q7. inventory = { &quot;The Great Gatsby&quot;: {&quot;price&quot;: 10.99, &quot;quantity&quot;: 5}, &quot;Rich dad poor dad&quot;: {&quot;price&quot;:
8.99, &quot;quantity&quot;: 8}, &quot;Harry Potter&quot;: {&quot;price&quot;: 12.50, &quot;quantity&quot;: 2}, }
• Add a new book in inventory using method and get all book titles using method
• Remove a book if it&#39;s out of stock

```python
inventory = { "The Great Gatsby": {"price": 10.99, "quantity": 5},
              "Rich dad poor dad": {"price": 8.99, "quantity": 8},
              "Harry Potter": {"price": 12.50, "quantity": 2}, } 

inventory.update({
    'Sam Altman':{"price":15.00,'quantity':5}
})
inventory.keys()
for book in list(inventory.keys()):
    if inventory[book]["quantity"] == 0:
        print(inventory.pop(book))
    
print(inventory)

```

Q10. Set Comparison

```python
A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7}
print(A.intersection(B))
print(A.difference(B))
print(B.difference(A))
```
