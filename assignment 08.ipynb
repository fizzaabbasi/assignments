Qno1: Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        print(f"Name: {self.name}")
        print(f"Number: {self.number}")
        print(f"Address: {self.address}")
        print(f"Salary: {self.salary}")
        print(f"Post: {self.post}")
        print(f"Year Of Joining: {self.year_of_joining}")
        print(f"Education: {self.education}")
        print(f"Year Of Experience: {self.year_of_experience}")

    def update_year_of_experience(self, new_year_of_experience):
        self.year_of_experience = new_year_of_experience
        print("Experience Updated To:", self.year_of_experience)

    def increase_salary(self, increment):
        self.salary += increment
        print("Increased Salary:", self.salary)

employee101 = Employee(
    "Fizza Tahseen",
    3138521880,
    "R172",
    50000,
    "Data Analyst",
    2026,
    "Undergraduate",
    1
)

print("Four Attributes Individually:")
print("Name:", employee101.name)
print("Number:", employee101.number)
print("Address:", employee101.address)
print("Salary:", employee101.salary)


print("\nComplete Employee Details:")
employee101.describe_employee()


print("\nAfter Updates:")
employee101.update_year_of_experience(2)
employee101.increase_salary(10000)


employee1 = Employee("Ali", 101, "Karachi", 50000, "Developer", 2023, "BSCS", 3)

employee2 = Employee("Sara", 102, "Lahore", 60000, "Designer", 2022, "BSIT", 4)

employee3 = Employee("Ahmed", 103, "Islamabad", 70000, "Manager", 2020, "MBA", 6)

employee4 = Employee("Ayesha", 104, "Karachi", 55000, "HR Officer", 2024, "BBA", 2)

employee5 = Employee("Usman", 105, "Multan", 45000, "Accountant", 2021, "B.Com", 5)

employee6 = Employee("Zainab", 106, "Quetta", 50000, "Teacher", 2023, "MA", 3)

employee7 = Employee("Hamza", 107, "Peshawar", 80000, "Team Lead", 2019, "BSCS", 7)

employee8 = Employee("Fatima", 108, "Hyderabad", 65000, "Data Analyst", 2022, "BSIT", 4)

employee9 = Employee("Bilal", 109, "Karachi", 58000, "Web Developer", 2021, "BSCS", 5)

employee10 = Employee("Mariam", 110, "Lahore", 75000, "Project Manager", 2019, "MBA", 7)



employees = [
    employee1,
    employee2,
    employee3,
    employee4,
    employee5,
    employee6,
    employee7,
    employee8,
    employee9,
    employee10
]


print("\n\nDetails of 10 Employees:")

for employee in employees:
    employee.describe_employee()
    print("------------------------------")



print("\nUpdates:")

employee1.increase_salary(5000)

employee2.update_year_of_experience(5)

employee3.increase_salary(10000)

employee4.update_year_of_experience(3)

employee5.increase_salary(7000)
```
