Question 1:Write a Python program that asks the user to guess a secret word.

Keep asking until the user enters the correct word ("python").

When guessed correctly, print "🎉 You guessed it!"
    

```python
word = 'python'
guess =""
while guess!= word:
    guess = input("Guess the word").lower()
    if guess == word:
        print("You guess it!!")
```

Question 2:Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
for i in users:
    print("Email send to: ",i)
    if i ==users[2]:
        break

print('Limit reached.Stop sending emails!')
    
```

Question 3: Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python
temps = [22, 25, 30, 45, 28] 
critical = 40
for temp in temps:
    if temp>=critical:
        print("Critical temperature reached!Stop monitoring,")
        break
        
    print("Temperature: ",temp)
```

Question 4: Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

```python
cart = [100, 250, 50, 300]
add = 0
for count in cart:
    add += count

print("Total Price:", add)
```

Question 5: Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True}
]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [ {"name": "Ali", "paid": True},
{"name": "Sara", "paid": False},
{"name": "John", "paid": True}
]
for customer in customers:
    if customer["paid"] == False:
        continue
    else:
        print("Access granted to: ",customer["name"])
        
```

Question 7: Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']



```python
domains=[]
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
for email in emails:
    domain = email.split("@")[1]
    domains.append(domain)

print("Domains: ",domains)
```

Question 8: Print only overdue tasks (Using for loop)

tasks = [ {"task": "Pay bills", "status": "done"},

{"task": "Submit report", "status": "pending"},

{"task": "Clean house", "status": "pending"}
]

Ouput:

Overdue: Submit report

Overdue: Clean house

```python
tasks = [ {"task": "Pay bills", "status": "done"},

{"task": "Submit report", "status": "pending"},

{"task": "Clean house", "status": "pending"}
]
for x in tasks:
    if x["status"] == "done":
        continue
    else:
        print("Overdue: ",x["task"])
```

Question 9:Write a Python program that keeps asking the user to enter numbers and prints them.

The program should stop the loop if the user enters a negative number.

Use a while True loop with a break statement.

```python
while True:
    num = int(input("Enter a number(negative to stop):"))
    print("You entered:",num)
    if num<0:
        print("Negative number entered.Loop stopped!")
        break
```

Question 10: Skip weekends while printing workdays (Using for loop and continue)

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]
for day in days:
    if day in ["Saturday","Sunday"]:
        continue
    else:
        print("Workday:",day)
```
