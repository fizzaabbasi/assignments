```python
import csv
```

```python
open_file = open("student data.csv", mode ="r")
open_file
```

```python
reader = csv.reader(open_file)
reader
```

```python
data = list(reader)
data 
```

```python
header = data[0]
header.append("Grade")
print(header)
```

```python
student_data = data[1:]
student_data
```

```python
for per in student_data:
    percentage = float(per[5])
    
    if percentage >=80:
        grade = "A"
    elif percentage >=70:
        grade = "B"
    elif percentage >=60:
        grade = "C"
    elif percentage >=50:
        grade = "D"
    else:
        grade = "F"

    per.append(grade)
    print(per)
```

```python
open_write = open("data.csv", mode = "w", newline="")
open_write
```

```python
writer = csv.writer(open_write)
writer
```

```python
writer.writerows(data)
```

```python
open_write.close()
```
